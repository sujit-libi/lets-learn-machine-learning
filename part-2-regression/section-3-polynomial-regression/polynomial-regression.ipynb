# Polynomial regression

### Intro
`y = b0 + b1x1 + b2x1^2 + b3x1^3 + ..... + bnx1^n`

#### Techniques for building a polynomial regression model for salary prediction (Stepwise)
1. Import the Libraries
2. Import the dataset
3. Training the Linear Regression model on the whole dataset
4. Training the Polynomial Regression model on the whole dataset
5. Visualizing the Linear Regression results
6. Visualizing the polynomial regression results
7. Visualizing the polynomial regression results (for higher resolution and smoother curve)
8. Predicting a new results with linear regression
9. Predicting a new results with Polynomial regression

### Step 1: Setting up data for linear vs polynomial regression comparison

```python
# Import the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Importing the dataset
dataset = pd.read_csv("Position_Salaries.csv")
X = dataset.iloc[:,1:-1].values
y = dataset.iloc[:,-1].values
```

### Step 2: linear to polynomial regression: Preparing data for advance model and transforming linear to polynomial

```python
# Training the linear Regression model on the whole dataset
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X, y)
```

```python
# Training the polynomial regression model o the whole dataset
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree = 5)
# degree(n): 2
# b0 + b1x1 + b2x1^2
X_poly = poly_reg.fit_transform(X)
lin_reg_2 = LinearRegression()
lin_reg_2.fit(X_poly, y)
```

### Step 3: Plotting Real v/s Predicted Salaries: Linear Regresision Visualization

```python
plt.scatter(X, y, color="red")
plt.plot(X, lin_reg.predict(X), color="blue")
plt.title("Truth or Bluff (Linear Regression)")
plt.xlabel("Position Level")
plt.ylabel("Salary")
plt.show()
```

### Step 4: Plotting Real vs Predicted Salaries: Polynomial Regression Visualization

```python
# Visualizing the Polynomial Linear Regression result with degree n: 2
plt.scatter(X, y, color="red")
plt.plot(X, lin_reg_2.predict(X_poly), color="blue")
plt.title("Truth or Bluff (Polynomial Regression)")
plt.xlabel("Position Level")
plt.ylabel("Salary")
plt.show()
```

```python
# Visualizing the polynomial linear regression result for higher resolution
X_grid = np.arange(X.min(), X.max(), 0.1)
X_grid = X_grid.reshape((len(X_grid),1))
plt.scatter(X, y, color="red")
plt.plot(X_grid, lin_reg_2.predict(poly_reg.fit_transform(X_grid)), color="blue")
plt.title("Truth or Bluff (Polynomial Regression)")
plt.xlabel("Position Level")
plt.ylabel("Salary")
plt.show()
```

### Step 5: Predicting Salaries - Linear Regression and Polynomial Regression

```python
# Predicting a new result with linear Regression
lin_reg.predict([[6.5]])
```

```python
# Predicting a new result with polynomial Regression
lin_reg_2.predict(poly_reg.fit_transform([[6.5]]))
```
