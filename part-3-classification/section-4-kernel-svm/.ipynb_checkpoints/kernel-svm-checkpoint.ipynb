# Kernel SVM
#### Terminology
- From linear to non-linear SVM: Linearly separate and non-linearly separate
- A higher dimension space
- mapping to a higher dimension
- But there is a catch: Mapping to a Higher Dimensional Space can be highly compute-intensive

### Types of Kernel Function
- Gaussian RBF Kernel
  $$\huge
    k(\vec{x}, \vec{l^i}) = e^{-\frac{\|\vec{x} - \vec{l^i}\|^2}{2\sigma^2}}
  $$
- Sigmoid Kernel
  $$\Large
      k(x,y) = tanh(\gamma x^Ty+ r)
  $$
- Polynomial Kernel
  $$\Large
      k(X,Y) = (\gamma*X_TY+r)^d , \gamma > 0
  $$

# Hands on

````python
# Importing the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
````

````python
# Importing the dataset
dataset = pd.read_csv("Social_Network_Ads.csv")
X = dataset.iloc[:,:-1].values
y = dataset.iloc[:,-1].values
````

````python
# Splitting the dataset into Training set and Test Set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=0)
````

````python
# Feature Scaling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
````

````python
# Training the Kernel SVM Model on the Training set
from sklearn.svm import SVC
classifier = SVC(kernel="rbf", random_state=0)
classifier.fit(X_train, y_train)
````

````python
# Predicting the new result
print(classifier.predict(sc.transform([[30,87000]])))
````

````python
# Predicting the test result
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
````

````python
# Making the Confusion Matrix and Calculating Accuracy Score
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_pred, y_test)
print(cm)
accuracy_score(y_pred, y_test)
````

**1. Simple Explanation**

A confusion matrix shows how many predictions your model got right vs. wrong, broken down by class. Your model got **93 out of 100** predictions correct — that's where the `0.93` accuracy comes from.

**2. Key Concepts**

| Term | Meaning |
|---|---|
| True Negative (TN) | Actual = No, Predicted = No (correct) |
| False Positive (FP) | Actual = No, Predicted = Yes (wrong — "false alarm") |
| False Negative (FN) | Actual = Yes, Predicted = No (wrong — "missed it") |
| True Positive (TP) | Actual = Yes, Predicted = Yes (correct) |

By default (scikit-learn's `confusion_matrix`), the layout is:
```
                Predicted 0   Predicted 1
Actual 0    [    TN      ,      FP      ]
Actual 1    [    FN      ,      TP      ]
```

**3. Step-by-Step Breakdown — Mapping Your Numbers**

```
[[64  3]
 [ 4 29]]
```

| | Predicted 0 | Predicted 1 |
|---|---|---|
| **Actual 0** | TN = 64 | FP = 3 |
| **Actual 1** | FN = 4 | TP = 29 |

- **64** — model correctly predicted "0" (e.g., no breach / benign) 64 times.
- **3** — model wrongly predicted "1" when it was actually "0" (false alarm).
- **4** — model wrongly predicted "0" when it was actually "1" (missed a real case).
- **29** — model correctly predicted "1" (e.g., breach/positive class) 29 times.

**Total samples:** 64+3+4+29 = **100**

**Accuracy check:**
$$\text{Accuracy} = \frac{TP + TN}{Total} = \frac{29 + 64}{100} = 0.93$$
✅ Matches your reported score.

**4. Example (applied to your cybersecurity context)**

If this were classifying "is this a breach (1) or not (0)":
- Out of 67 actual non-breaches, model correctly flagged 64, wrongly flagged 3 as breaches.
- Out of 33 actual breaches, model correctly caught 29, but **missed 4 real breaches** (these are the dangerous ones).

**5. Real-World Application**

In security, **False Negatives (4 here)** are usually the scariest — they mean real breaches slipped through undetected. **False Positives (3 here)** just mean extra alerts to review — annoying but less dangerous. This is why in security ML models, teams often care more about **Recall** than plain Accuracy.

**6. Common Mistakes**
- Assuming high accuracy alone means a "good" model — with imbalanced classes (e.g., 95% non-breach, 5% breach), a model could get 95% accuracy just by *always* predicting "no breach" and never catching real ones.
- Mixing up rows vs. columns — always confirm which axis is Actual vs. Predicted for the specific library/version you're using.

**Quick Summary**

- 93/100 correct → 0.93 accuracy ✔️
- 3 false alarms, 4 missed real cases
- For security use cases, also check **Precision**, **Recall**, and **F1-score** — not just accuracy — since missed breaches (FN) matter more than false alarms (FP).

**Beginner Notes:** Always compute Precision = TP/(TP+FP) and Recall = TP/(TP+FN) alongside accuracy — here Precision = 29/32 ≈ 0.906, Recall = 29/33 ≈ 0.879.

**Exam Tip:** Memorize the formulas: Accuracy = (TP+TN)/Total, Precision = TP/(TP+FP), Recall = TP/(TP+FN), F1 = 2×(Precision×Recall)/(Precision+Recall) — these are guaranteed exam/interview questions.

**Next Topic to Learn:** Precision-Recall tradeoff and F1-score — critical for evaluating models on imbalanced datasets like fraud/breach detection, where accuracy alone can be misleading.

````python
# Visualising the Training set results
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_train), y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 1),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 1))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('K-NN (Training set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
````

````python
# Visualising the Test set results
from matplotlib.colors import ListedColormap
X_set, y_set = sc.inverse_transform(X_test), y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 10, stop = X_set[:, 0].max() + 10, step = 1),
                     np.arange(start = X_set[:, 1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 1))
plt.contourf(X1, X2, classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('K-NN (Test set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
````
