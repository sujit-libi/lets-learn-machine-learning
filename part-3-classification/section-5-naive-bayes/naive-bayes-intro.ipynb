# Naive Bayes
## Intro
### Bayes Theorem

$
    P(A|B) = \frac{P(B|A) * P(A)}{P(B)}
$

#### For example
- Machine_one = 30 wrenches / hrs
- Machine_two = 20 wrenches / hrs

- Out of all produced parts: We can see that `1%` are defective
- Out of all defective parts: We can see that `50%` came from machine_one and `50%` came from machine_two

**Question**: What is the probability that a part produced by machine_two is defective?

**Solution**

*Given:*
- P(machine_one) = 30/50 = 0.6
- P(machine_two) = 20/50 = 0.4
- P(defect) = 1% = 0.01
- P(machine_one | defect) = 50% = 0.5
- P(machine_two | defect) = 50% = 0.5
- P(defect | machine_two) = ?

$
P(defect | machine\_two) = \frac{P(machine\_two | defect) * P(defect)}{P(machine\_two)}
$

$
p(defect | machine\_two) = \frac{0..5 * 0.01}{0.4}
$

$
p(defect | machine\_two) = 0.0125
$

$
p(defect | machine\_two) = 1.25%
$

#### Lets look into some practice example for classification
**Assume**

- `X1` => Age => X-axis
- `X2` => Salary => y-axis
- `Category_1` => triangle shape as Walks
- `Category_2` => circle shape as Drives
- Trying to classify person if they are more into walks or drive based on their age and salary
- Use Naive Bayes

**Available Information**
- Total No. of observation => 30
- Number of Walkers => 10
- Number of Drivers => 20

#### Some Terminology
- P(Walks) : Prior Probability => 1
- P(X) : Marginal likelihood => 2
- P(X | Walks) : Likelihood => 3
- P(Walks | X) : Posterior Probability => 4

*Formula*

$
 P(Walks | X) = \frac{P(X | Walks) * P(Walks)}{P(X)}
$

## For Walks
#### Step 1: Prior Probability => P(Walks)
- P(Walks) = `Number of Walkers / Total no. of observation`
- P(Walks) = `10/30`
- P(Walks) = `0.333`

#### Step 2: Marginal Likelihood => P(X)
- P(X) = `Number of similar observation / Total no. of observation`
- P(X) = `4/30`
- P(X) = `0.133`
  
`Note: Number of similar observation => observation inside circle => we create circle by setting radius from a new data point`

#### Step 3: Likelihood => P(X | Walks)
-  $ P(X | Walks) = \large \frac{Number of similar observations among those who walk}{Total no. of walkers} $
-  $ P(X | Walks) = 3/10 $
-  $ P(X | Walks) = 0.3$
  
`Note: Number of similar observations among those who walks => observation of walks inside circle which we create circle by setting radius from a new data point earlier`

#### Step 4: Posterior Probability
- $ P(Walks | X) = \frac{P(X | Walks) * P(Walks)}{P(X)} $
- $ P(Walks | X) = \frac{0.3 * 0.333}{0.133} $
- $ P(Walks | X) = 0.75 $
- $ P(Walks | X) = 75% $



## Similarly For Drives
#### Step 1: Prior Probability => P(Drives)
- P(Drives) = `Number of Drivers / Total no. of observation`
- P(Drives) = `20/30`
- P(Drives) = `0.6667`

#### Step 2: Marginal Likelihood => P(X)
- P(X) = `Number of similar observation / Total no. of observation`
- P(X) = `4/30`
- P(X) = `0.133`
  
`Note: Number of similar observation => observation inside circle => we create circle by setting radius from a new data point`

#### Step 3: Likelihood => P(X | Drives)
-  $ P(X | Drives) = \large \frac{Number of similar observations among those who drive}{Total no. of drivers} $
-  $ P(X | Drives) = 1/20 $
-  $ P(X | Drives) = 0.05$
  
`Note: Number of similar observations among those who drives => observation of drives inside circle which we create circle by setting radius from a new data point earlier`

#### Step 4: Posterior Probability
- $ P(Drives | X) = \frac{P(X | Drives) * P(Drives)}{P(X)} $
- $ P(Drives | X) = \frac{0.05 * 0.6667}{0.133} $
- $ P(Drives | X) = 0.25 $
- $ P(Drives | X) = 25% $
