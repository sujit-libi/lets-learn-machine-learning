# Classification Intro
- Machine Learning technique to identify the category of new observations based on training data
- For Example:
  1. Likely to Stay or Likely to Leave
  2. Dogs or Cats
  3. Email - Important, Urgent or Spam

# Logistic Regression
- Predict a categorical dependent variable from a number of independent variables
- like : will purchase, health insurance: yes/no

  $ln \frac{p}{1-p} = b_0 + b_1 * x_1$

    $ln \frac{p}{1-p} = b_0 + b_1 * x_1 + b_2 * x_2 + ..... + b_n * x_n$

# Hands-On: Logistic Regression Steps
- Importing the libraries
- Importing the dataset
- Splitting the dataset into the training set and test set
- Feature Scaling
- Training the Logistic Regression Model on the Training set
- Predicting a new result
- Making the confusion matrix
- Visualizing the Training set result
- Visualizing the Test set result


```python
# Importing the libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Importing the dataset
dataset = pd.read_csv("Social_Network_Ads.csv")
X = dataset.iloc[:,:-1].values
y = dataset.iloc[:,-1].values
```

```python
# Splitting the dataset into training set and test set
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, random_state=0)
```

```python
# Feature Scalling
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
# Logistic Regression
from sklearn.linear_model import LogisticRegression
classifier = LogisticRegression(random_state = 0)
```

```python
classifier.fit(X_train, y_train)
```

```python
# Predicting
classifier.predict([[19, 20000]])
classifier.predict([[30, 87000]])
```

```python
# Predict new result
classifier.predict(sc.transform([[30, 87000]]))
print(classifier.predict(sc.transform([[30, 87000]])))
```

```python
# Comparing predicted vs real results
y_pred = classifier.predict(X_test)
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```python
# Implementing Confusion Matrix and Accuracy Score in Scikit-learn
from sklearn.metrics import confusion_matrix, accuracy_score
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
```

```python
# Visualizing the training set result for logistic regression decision boundaries
from matplotlib.colors import ListedColormap

X_set, y_set = sc.inverse_transform(X_train), y_train
X1, X2 = np.meshgrid(
    np.arange(start = X_set[:,0].min() - 10, stop = X_set[:,0].max() + 10, step = 0.25), 
    np.arange(start = X_set[:,1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 0.25)
)
plt.contourf(
    X1, X2, 
    classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape), 
    alpha = 0.75, cmap = ListedColormap(("red", "green"))
)

plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], 
                c=ListedColormap(("red","green"))(i), label=j)

plt.title("Logistic Regression (Training Set)")
plt.xlabel("Age")
plt.ylabel("Estimated Salary")
plt.legend()
plt.show()
```

```python
# Visualizing the test set result for logistic regression decision boundaries
from matplotlib.colors import ListedColormap

X_set, y_set = sc.inverse_transform(X_test), y_test
X1, X2 = np.meshgrid(
    np.arange(start = X_set[:,0].min() - 10, stop = X_set[:,0].max() + 10, step = 0.25), 
    np.arange(start = X_set[:,1].min() - 1000, stop = X_set[:, 1].max() + 1000, step = 0.25)
)
plt.contourf(
    X1, X2, 
    classifier.predict(sc.transform(np.array([X1.ravel(), X2.ravel()]).T)).reshape(X1.shape), 
    alpha = 0.75, cmap = ListedColormap(("red", "green"))
)

plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())

for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1], 
                c=ListedColormap(("red","green"))(i), label=j)

plt.title("Logistic Regression (Test Set)")
plt.xlabel("Age")
plt.ylabel("Estimated Salary")
plt.legend()
plt.show()
```
