### Step 1: Machine Learning Toolkit: Importing Numpy, Matplotlib, and pandas libraries

```python
# Importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### Step 2: Machine Learning Basics: Importing Datasets using Pandas read_csv()

```python
# Importing the dataset
dataset = pd.read_csv("Data.csv")
# Using pandas iloc for feature selection in ML data preprocessing
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:,-1].values
```

```python
# Preprocessing Data: Building X and Y Vectors for ML Model Training
print("""
# ----------------------------------------------------------------- #
# X
# ----------------------------------------------------------------- #
""")
print(X)

print("""
# ----------------------------------------------------------------- #
# y
# ----------------------------------------------------------------- #
""")
print(y)
```

### Step 3: Using Scikit-Learn to replace missing values in machine learning 

```python
# Taking care of missing values
from sklearn.impute import SimpleImputer
imputer = SimpleImputer(missing_values=np.nan, strategy="mean")

# Imputing missing data in python: SimpleImputer and Numerical Columns
imputer.fit(X[:,1:3])
X[:,1:3] = imputer.transform(X[:,1:3])

print(X)
```

### Step 4: One-Hot encoding: Transforming categorical features for Machine Learning

#### Encoding the independent variable

```python
# importing required libraries
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
```

```python
# Handling Categorical Data: One-hot Encoding with Column Transformer
ct = ColumnTransformer(transformers=[("encoder", OneHotEncoder(), [0])], remainder="passthrough")
print(ct.fit_transform(X))
```

```python
# Preprocessing categorical data: One hot and label encoding Technique
X = np.array(ct.fit_transform(X))
print(X)
```

#### Encoding the dependent variable

```python
from sklearn.preprocessing import LabelEncoder
lc = LabelEncoder()
y = lc.fit_transform(y)
print(y)
```

### Step 5: How to prepare Data for Machine Learning: Training and Test Sets

```python
# Splitting the data set into the training set and test set
# Q: do we need to train our dataset before or after feature scaling
# A: After: To prevent information leakage of test set before training is done

# Creating Training and Test sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=1)

# Splitting data into Training and Test sets: 
print("X train data")
print(X_train)
print("X test data")
print(X_test)
print("y train data")
print(y_train)
print("y test data")
print(y_test)
```

### Step 6: Feature Scaling in Machine Learning: Why its crucial for Data preprocessing
1. Standardisation

2. Normalization

```python
# How to Scale numeric features for machine learning
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()

# Implementing Feature Scaling: Fit and Transform methods
X_train[:,3:] = sc.fit_transform(X_train[:,3:])

# Applying the same scaler to training and test set
X_test[:, 3:] = sc.transform(X_test[:,3:])
print(X_train)
print(X_test)
```
